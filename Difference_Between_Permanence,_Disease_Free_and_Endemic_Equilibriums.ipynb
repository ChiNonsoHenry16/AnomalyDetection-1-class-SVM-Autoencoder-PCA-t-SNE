1. Define the Time Range

I created a range of time values
𝑡
t from 0 to 50 (with 500 points in between) to simulate how infection levels change over time:

t = np.linspace(0, 50, 500)

2. Model the Endemic Equilibrium

For endemic equilibrium, the infection level doesn’t change once it stabilizes.
I represented this as a constant line at a positive prevalence (e.g., 0.4):

endemic_equilibrium = np.ones_like(t) * 0.4


This means the infection is always at 40% of the population (or network nodes).

3. Model Virus Permanence

For permanence, infection persists but fluctuates over time.
I used a sine wave to simulate oscillations:

permanence = 0.3 + 0.15 * np.sin(0.6 * t)


Baseline = 0.3 (infection stays around this level).

Amplitude = 0.15 (fluctuations up and down).

Frequency = 0.6 (controls how fast it oscillates).

To make sure the infection never drops to zero, I set a lower bound:

permanence[permanence < 0.05] = 0.05


So even at the lowest oscillation, the infection stays above 0.05 (i.e., it never disappears).

4. Plot Both Curves

I then plotted:

A solid blue line for endemic equilibrium.

A red dashed line for permanence.
I added axis labels, a title, and a legend for clarity:

```python
import matplotlib.pyplot as plt
import numpy as np

# Time range
t = np.linspace(0, 50, 500)

# Example of endemic equilibrium: infection settles at a constant positive level
endemic_equilibrium = np.ones_like(t) * 0.4

# Example of permanence: infection fluctuates but stays above zero
permanence = 0.3 + 0.15 * np.sin(0.6 * t)  # oscillatory permanence
permanence[permanence < 0.05] = 0.05  # keep it bounded away from zero

# Example of disease-free equilibrium: infection goes to zero
disease_free_equilibrium = np.zeros_like(t)

# Plotting with thicker lines
plt.figure(figsize=(10, 6))
plt.plot(t, endemic_equilibrium, label="Endemic Equilibrium (constant prevalence)", color="blue", linewidth=4)
plt.plot(t, permanence, label="Permanence (fluctuating prevalence)", color="red", linewidth=4, linestyle="--")
plt.plot(t, disease_free_equilibrium, label="Disease-Free Equilibrium", color="green", linewidth=4, linestyle=":")

# Labels and title
plt.xlabel("Time", fontsize=12)
plt.ylabel("Infection Level (I)", fontsize=12)
plt.title("Endemic Equilibrium, Permanence, and Disease-Free Equilibrium", fontsize=14)
plt.legend()
plt.grid(True)
plt.show()
```
